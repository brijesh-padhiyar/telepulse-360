### 2️⃣ Step 2: Load & Inspect the 14-Column Master Data Schema

```python
# Locate schema file relative to notebook directory or project root
possible_paths = [
    Path("../data/data_schema.json"),
    Path("data/data_schema.json"),
    Path("e:/Venom/telepulse-360/data/data_schema.json")
]

schema_file = next((p for p in possible_paths if p.exists()), None)
assert schema_file is not None, "❌ Error: data/data_schema.json not found!"

with open(schema_file, "r", encoding="utf-8") as f:
    schema_data = json.load(f)

print(f"✅ Project Name   : {schema_data.get('project_name')}")
print(f"🎯 Target Scale   : {schema_data.get('target_records'):,} Transactions")
print(f"📊 Total Fields   : {len(schema_data.get('schema', []))} Columns")
```

```python
# Render Data Dictionary as a Clean Tabular View
try:
    import pandas as pd
    df_schema = pd.DataFrame(schema_data.get("schema", []))
    df_schema = df_schema[["column", "type", "description"]]
    df_schema.columns = ["Field Name", "Data Type", "Telecom Business Description"]
    display(df_schema)
except ImportError:
    print("Pandas loading... Printing raw columns:")
    for col in schema_data.get("schema", []):
        print(f" - {col['column']:<22} | {col['type']:<12} | {col['description']}")
```
